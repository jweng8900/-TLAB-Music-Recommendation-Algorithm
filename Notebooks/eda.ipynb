```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../Data/train.csv")
```

```python
df.head()
```

```python
df.info()
df.isnull().sum()
```

```python
lyrical_cols = [
    'dating', 'violence', 'world/life', 'night/time', 'shake the audience', 'family/gospel',
    'romantic', 'communication', 'obscene', 'music', 'movement/places', 
    'light/visual perceptions', 'family/spiritual', 'sadness', 'feelings', 'age'
]
```

```python
df[lyrical_cols].describe()
```

```python
df[lyrical_cols].hist(figsize=(15, 10), bins=30)
plt.suptitle('Distributions of Lyrical Scores')
plt.tight_layout()
```

```python
plt.figure(figsize=(14, 10))
sns.heatmap(df[lyrical_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Lyrical Features')
plt.show()
```

```python
df['genre'].value_counts().head(10).plot(kind='barh', title='Top Genres')

genre_means = df.groupby('genre')[lyrical_cols].mean()
genre_means.T.plot(kind='bar', figsize=(18, 6), title='Average Lyrical Scores by Genre')
plt.ylabel('Score')
```

```python
print(df['release_date'].isna().sum(), "missing release dates")
print(df['release_year'].value_counts().sort_index())
```

```python
df[['release_date']].head(10)
```

```python
# Extract year from the weird timestamp string
df['release_year'] = df['release_date'].astype(str).str[-4:].astype(int)

# Now group and plot
songs_per_year = df['release_year'].value_counts().sort_index()

plt.figure(figsize=(12, 6))
sns.lineplot(x=songs_per_year.index, y=songs_per_year.values)
plt.title('Number of Songs Released by Year')
plt.xlabel('Year')
plt.ylabel('Number of Songs')
plt.grid(True)
plt.show()
```
