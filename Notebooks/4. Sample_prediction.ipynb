```python
# 1. Imports
import pandas as pd
import joblib

# 2. Load the full preprocessing+clustering pipeline
pipeline = joblib.load("models/full_pipeline.joblib")

# 3. Read in the new samples
df_test = pd.read_csv("../Data/recommend.csv")
print("Original test shape:", df_test.shape)

# 4. Drop unused columns
drop_cols = [
    "Unnamed: 0", "lyrics", "artist_name", "track_name",
    "release_date", "genre", "topic"
]
df_clean = df_test.drop(columns=drop_cols, errors="ignore")

# 5. Define the numeric columns exactly as in your training pipeline
train_numeric_cols = [
    'len',
    'violence',
    'world/life',
    'night/time',
    'romantic',
    'communication',
    'obscene',
    'music',
    'sadness'
]

# 6. Predict clusters in one go
df_test["cluster_km"] = pipeline.predict(df_clean[train_numeric_cols])

# 7. Inspect and save
print("\nCluster distribution on new data:")
print(df_test["cluster_km"].value_counts())

out_path = "../Data/recommend_with_clusters.csv"
df_test.to_csv(out_path, index=False)
print(f"\nSaved labeled recommendations to {out_path}")


```
